# Create a Balanced dataset

```python
import os
import random
import shutil

# Paths
original_dataset_path = r"C:\Users\acer\Desktop\Waste detection and classification\archive\garbage-dataset"  # Replace with the path to your original dataset
new_dataset_path = r"C:\Users\acer\Desktop\Waste detection and classification\archive\garbage-dataset/new_dataset"  # Replace with the path for the new dataset

# Categories and number of images to sample
categories = ["Metal", "Glass", "Biological", "Paper", "Cardboard", "Plastic"]
sample_size = 400

# Create a new dataset folder
if not os.path.exists(new_dataset_path):
    os.makedirs(new_dataset_path)

# Loop through each category and sample images
for category in categories:
    category_path = os.path.join(original_dataset_path, category)
    new_category_path = os.path.join(new_dataset_path, category)

    # Create a folder for the category in the new dataset
    os.makedirs(new_category_path, exist_ok=True)

    # Get all images from the category folder
    all_images = [img for img in os.listdir(category_path) if os.path.isfile(os.path.join(category_path, img))]

    # Randomly sample images
    sampled_images = random.sample(all_images, min(sample_size, len(all_images)))

    # Copy sampled images to the new dataset
    for img in sampled_images:
        src_path = os.path.join(category_path, img)
        dest_path = os.path.join(new_category_path, img)
        shutil.copy(src_path, dest_path)

    print(f"Sampled {len(sampled_images)} images from {category}.")

print("Dataset creation completed!")
```
